## Implementasi From Scratch

Sesuai aturan kelompok, semua kode from scratch ditulis langsung di notebook ini, bukan di file Python terpisah. Bagian ini berisi model Logistic Regression dan fungsi pembagian bin dan fungsi bantu validasi yang dipakai di bagian-bagian berikutnya.

```python
# Sigmoid yang stabil secara numerik sehingga tidak overflow untuk z yang sangat besar atau sangat kecil
# Fungsi ini memakai modul array yang sama dengan z sehingga bisa berjalan di NumPy maupun CuPy
def sigmoid(z):
    xp = _array_module(z)
    e = xp.exp(-xp.abs(z))
    return xp.where(z >= 0, 1.0 / (1.0 + e), e / (1.0 + e))


# Ambil modul array dari sebuah array, yaitu cupy untuk array GPU dan numpy untuk lainnya
def _array_module(a):
    if type(a).__module__.split('.')[0] == 'cupy':
        import cupy
        return cupy
    return np


# Pilih modul array untuk training dengan 'cpu' memakai NumPy dan 'gpu' memakai CuPy
# CuPy adalah pustaka array dengan sintaks seperti NumPy yang berjalan di GPU NVIDIA
def _select_backend(device):
    if device not in ('cpu', 'gpu'):
        raise ValueError(f"device tidak dikenal: {device}")
    if device == 'cpu':
        return np
    try:
        import cupy
        if cupy.cuda.runtime.getDeviceCount() > 0:
            return cupy
        error = 'tidak ada GPU NVIDIA yang terdeteksi'
    except Exception as e:
        error = f'CuPy tidak bisa dipakai ({type(e).__name__})'
    raise RuntimeError(f"device='gpu' diminta tetapi {error}. Pasang CuPy (misalnya 'pip install cupy-cuda13x[ctk]') "
                       f"atau pakai device='cpu'.")


# Salin array NumPy atau CuPy ke memori CPU sebagai array NumPy float64
def _to_host(a):
    return np.asarray(a.get() if hasattr(a, 'get') else a, dtype=float)


# Logistic Regression berbasis gradien dengan regularisasi L1, L2, dan Elastic Net
# Fungsi objektif disamakan dengan scikit-learn agar nilai C bisa dibandingkan langsung
#   J = C * sum(s_i * logloss_i) + (1 - rho) / 2 * ||w||^2 + rho * ||w||_1
# Objektif tersebut dibagi C * sum(s) agar skala gradien tidak bergantung pada jumlah data
# s_i adalah bobot sampel dan rho bernilai 0 untuk L2, 1 untuk L1, dan l1_ratio untuk Elastic Net
# Bias tidak ikut diregularisasi
# Sumber FISTA adalah Beck dan Teboulle 2009 dalam SIAM Journal on Imaging Sciences
# Sumber adaptive restart adalah O'Donoghue dan Candes 2015 dalam Foundations of Computational Mathematics
# Inovasi di luar gradient descent biasa adalah FISTA, adaptive restart, langkah 1 / L, dan proximal step untuk L1
# Inovasi lain adalah bobot kelas untuk cost-sensitive dan backend GPU lewat CuPy
class LogisticRegression:

    # penalty bernilai None, 'l1', 'l2', atau 'elasticnet'
    # C adalah kebalikan kekuatan regularisasi dan l1_ratio adalah porsi L1 pada Elastic Net
    # solver bernilai 'gd' untuk gradient descent biasa atau 'fista'
    # learning_rate berupa angka atau 'auto' untuk 1 dibagi konstanta Lipschitz
    # tol menghentikan training jika perubahan parameter terbesar dalam satu iterasi lebih kecil dari nilai ini
    # class_weight bernilai None atau dict {kelas: bobot} yang dipakai untuk cost-sensitive
    # record_params menyimpan bias dan bobot di setiap iterasi untuk visualisasi lintasan
    # device bernilai 'cpu' untuk NumPy atau 'gpu' untuk CuPy
    # Hanya training yang berjalan di GPU dan hasilnya (coef_, intercept_) disalin ke CPU
    # dtype bernilai 'float64' atau 'float32' dan float32 lebih cepat di GPU tetapi hasilnya tidak persis sama
    def __init__(self, penalty='l2', C=1.0, l1_ratio=0.5, solver='fista', learning_rate='auto',
                 max_iter=5000, tol=1e-5, class_weight=None, record_params=False, device='cpu',
                 dtype='float64'):
        self.penalty = penalty
        self.C = C
        self.l1_ratio = l1_ratio
        self.solver = solver
        self.learning_rate = learning_rate
        self.max_iter = max_iter
        self.tol = tol
        self.class_weight = class_weight
        self.record_params = record_params
        self.device = device
        self.dtype = dtype

    # Latih bobot dan bias, nama kolom disimpan di feature_names_ jika X berupa DataFrame
    # Kedua solver memakai langkah yang sama yaitu gradien log-loss dan L2 lalu proximal step untuk L1
    # Bedanya FISTA menghitung gradien di titik momentum Nesterov sehingga konvergen lebih cepat pada fitur berkorelasi
    def fit(self, X, y):
        if self.solver not in ('gd', 'fista'):
            raise ValueError(f'solver tidak dikenal: {self.solver}')
        if self.dtype not in ('float64', 'float32'):
            raise ValueError(f'dtype tidak dikenal: {self.dtype}')
        xp = _select_backend(self.device)
        self.feature_names_ = list(X.columns) if hasattr(X, 'columns') else None
        # Data disalin ke perangkat sekali di awal dan seluruh iterasi berjalan di perangkat yang sama
        X = xp.asarray(np.asarray(X, dtype=self.dtype))
        y = xp.asarray(np.asarray(y).astype(self.dtype))

        s = self._sample_weights(xp, y)
        S = float(s.sum())
        lam1, lam2 = self._penalty_strength(S)
        lr = self._step_size(xp, X, s, S, lam2)

        # (w, b) adalah parameter sekarang dan (wv, bv) adalah titik tempat gradien dihitung
        # z dan zv disimpan agar X @ w cukup dihitung sekali per iterasi
        w, b = xp.zeros(X.shape[1], dtype=X.dtype), 0.0
        wv, bv = w.copy(), b
        z = xp.zeros(len(y), dtype=X.dtype)
        zv = z.copy()
        t = 1.0

        self.loss_history_ = [self._objective(xp, z, y, s, S, w, lam1, lam2)]
        self.param_history_ = [np.r_[b, _to_host(w)]] if self.record_params else None
        self.converged_ = False

        for it in range(1, self.max_iter + 1):
            err = s * (sigmoid(zv) - y) / S
            w_new = wv - lr * (X.T @ err + lam2 * wv)
            b_new = bv - lr * float(err.sum())

            # Bagian L1 tidak terdiferensiasi di nol sehingga dipakai soft-thresholding (proximal step)
            # Dengan begitu bobot yang tidak penting benar-benar bernilai nol
            if lam1 > 0:
                w_new = xp.sign(w_new) * xp.maximum(xp.abs(w_new) - lr * lam1, 0.0)

            z_new = X @ w_new + b_new
            delta = max(float(xp.abs(w_new - w).max()), abs(b_new - b))

            if self.solver == 'fista':
                # Momentum di-reset jika arah langkah berbalik terhadap momentum (adaptive restart)
                # Reset ini mencegah parameter berosilasi di sekitar titik optimum
                if float((wv - w_new) @ (w_new - w)) + (bv - b_new) * (b_new - b) > 0:
                    t = 1.0
                t_next = (1.0 + np.sqrt(1.0 + 4.0 * t * t)) / 2.0
                beta = (t - 1.0) / t_next
                t = t_next
            else:
                beta = 0.0

            wv = w_new + beta * (w_new - w)
            bv = b_new + beta * (b_new - b)
            zv = z_new + beta * (z_new - z)
            w, b, z = w_new, b_new, z_new

            self.loss_history_.append(self._objective(xp, z, y, s, S, w, lam1, lam2))
            if self.record_params:
                self.param_history_.append(np.r_[b, _to_host(w)])
            if delta < self.tol:
                self.converged_ = True
                break

        self.coef_ = _to_host(w)
        self.intercept_ = float(b)
        self.n_iter_ = it
        self.learning_rate_ = lr
        if self.record_params:
            self.param_history_ = np.array(self.param_history_)
        return self

    # Nilai z = Xw + b sebelum masuk sigmoid
    def decision_function(self, X):
        return np.asarray(X, dtype=float) @ self.coef_ + self.intercept_

    # Probabilitas kelas 1
    def predict_proba(self, X):
        return sigmoid(self.decision_function(X))

    # Label 0 atau 1 berdasarkan threshold
    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)

    # Bobot tiap sampel dari class_weight untuk cost-sensitive dengan nilai 1 untuk kelas yang tidak disebut
    def _sample_weights(self, xp, y):
        if self.class_weight is not None and not isinstance(self.class_weight, dict):
            raise ValueError('class_weight harus None atau dict {kelas: bobot}')
        s = xp.ones(len(y), dtype=y.dtype)
        for c, wc in (self.class_weight or {}).items():
            s[y == c] *= wc
        return s

    # Kekuatan regularisasi L1 dan L2 setelah objektif dibagi C * sum(s)
    def _penalty_strength(self, S):
        if self.penalty is None:
            return 0.0, 0.0
        rho = {'l1': 1.0, 'l2': 0.0, 'elasticnet': self.l1_ratio}.get(self.penalty)
        if rho is None:
            raise ValueError(f"penalty tidak dikenal: {self.penalty}")
        scale = 1.0 / (self.C * S)
        return rho * scale, (1.0 - rho) * scale

    # Learning rate 'auto' adalah 1 / L dengan L konstanta Lipschitz gradien bagian yang halus
    # Turunan kedua sigmoid paling besar 0,25 sehingga L <= 0,25 * eig_max(X'SX / sum(s)) + lam2
    # Dengan langkah ini gradient descent dijamin turun tanpa mencoba learning rate satu per satu
    def _step_size(self, xp, X, s, S, lam2):
        if self.learning_rate != 'auto':
            return float(self.learning_rate)
        Xb = xp.hstack([xp.ones((len(X), 1), dtype=X.dtype), X])
        hessian_bound = 0.25 * (Xb.T * s) @ Xb / S
        return 1.0 / (float(xp.linalg.eigvalsh(hessian_bound).max()) + lam2)

    # Nilai fungsi objektif yaitu rata-rata log-loss berbobot ditambah penalti
    # log(1 + e^z) - y * z sama dengan log-loss dan tidak menghasilkan log(0)
    def _objective(self, xp, z, y, s, S, w, lam1, lam2):
        logloss = float(xp.sum(s * (xp.logaddexp(0.0, z) - y * z))) / S
        return logloss + lam1 * float(xp.abs(w).sum()) + 0.5 * lam2 * float(w @ w)
```

```python
# Cari batas bin satu fitur dengan pemisahan biner berulang yang paling menurunkan impuritas Gini terhadap target
# Pemisahan terbaik dipilih lebih dulu (best-first) sampai terbentuk n_bins kelompok dengan tiap kelompok minimal min_leaf baris
# Idenya diadaptasi dari kriteria pemisahan CART oleh Breiman dkk 1984 dalam Classification and Regression Trees
# Inovasi ini adalah binning terawasi yang menaruh batas bin di tempat rasio batal benar-benar berubah
def target_edges(x, t, n_bins, min_leaf):
    order = np.argsort(x, kind='mergesort')
    xs, ts = x[order], t[order].astype(float)
    cum = np.concatenate([[0.0], np.cumsum(ts)])

    # Pemisahan terbaik pada segmen [lo, hi) dengan hasil berupa (penurunan impuritas, posisi potong)
    def best_split(lo, hi):
        if hi - lo < 2 * min_leaf:
            return None
        cut = np.arange(lo + min_leaf, hi - min_leaf + 1)
        cut = cut[xs[cut - 1] != xs[np.minimum(cut, len(xs) - 1)]]
        if len(cut) == 0:
            return None
        n_l, n_r = cut - lo, hi - cut
        pos_l, pos_r = cum[cut] - cum[lo], cum[hi] - cum[cut]
        pos, n = cum[hi] - cum[lo], hi - lo
        # n * Gini / 2 = pos * (n - pos) / n
        after = pos_l * (n_l - pos_l) / n_l + pos_r * (n_r - pos_r) / n_r
        gain = pos * (n - pos) / n - after
        i = int(np.argmax(gain))
        return float(gain[i]), int(cut[i])

    segments = {(0, len(xs)): best_split(0, len(xs))}
    edges = []
    while len(edges) < n_bins - 1:
        open_ = {k: v for k, v in segments.items() if v is not None}
        if not open_:
            break
        (lo, hi), (_, cut) = max(open_.items(), key=lambda kv: kv[1][0])
        edges.append((xs[cut - 1] + xs[cut]) / 2.0)
        del segments[(lo, hi)]
        segments[(lo, cut)] = best_split(lo, cut)
        segments[(cut, hi)] = best_split(cut, hi)
    return np.array(sorted(edges))
```

```python
# Simpan objek model ke file pickle
def save_model(model, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    with open(path, 'wb') as f:
        pickle.dump(model, f)


# Muat kembali model yang disimpan dengan save_model
def load_model(path):
    with open(path, 'rb') as f:
        return pickle.load(f)


# Confusion matrix biner dalam bentuk [[TN, FP], [FN, TP]]
def confusion_matrix(y_true, y_pred):
    y_true = np.asarray(y_true).astype(int)
    y_pred = np.asarray(y_pred).astype(int)
    tp = int(np.sum((y_true == 1) & (y_pred == 1)))
    tn = int(np.sum((y_true == 0) & (y_pred == 0)))
    fp = int(np.sum((y_true == 0) & (y_pred == 1)))
    fn = int(np.sum((y_true == 1) & (y_pred == 0)))
    return np.array([[tn, fp], [fn, tp]])


# F1 untuk kelas 0 dan kelas 1, hasilnya [f1_0, f1_1]
def f1_per_class(y_true, y_pred):
    (tn, fp), (fn, tp) = confusion_matrix(y_true, y_pred)
    f1_pos = 2 * tp / (2 * tp + fp + fn) if tp + fp + fn > 0 else 0.0
    f1_neg = 2 * tn / (2 * tn + fn + fp) if tn + fn + fp > 0 else 0.0
    return np.array([f1_neg, f1_pos])


# Macro F1, rata-rata F1 kedua kelas (metrik resmi kompetisi)
def macro_f1(y_true, y_pred):
    return float(f1_per_class(y_true, y_pred).mean())


# Cari threshold dengan Macro F1 tertinggi, dipakai pada prediksi out-of-fold
def best_threshold(y_true, proba, grid=None):
    if grid is None:
        grid = np.round(np.arange(0.20, 0.81, 0.01), 2)
    scores = np.array([macro_f1(y_true, (proba >= t).astype(int)) for t in grid])
    i = int(np.argmax(scores))
    return float(grid[i]), float(scores[i])


# Validasi silang yang mengembalikan probabilitas out-of-fold
# Preprocessor di-fit ulang tiap fold hanya dengan data train fold tersebut
def cross_val_proba(make_model, make_prep, df, y, folds):
    y = np.asarray(y)
    oof = np.zeros(len(y))
    for tr_idx, va_idx in folds:
        prep = make_prep()
        X_tr = prep.fit_transform(df.iloc[tr_idx])
        X_va = prep.transform(df.iloc[va_idx])
        model = make_model()
        model.fit(X_tr, y[tr_idx])
        oof[va_idx] = model.predict_proba(X_va)
    return oof


# Macro F1 per fold dengan satu threshold, untuk melihat kestabilan model
def fold_scores(y, proba, folds, threshold):
    y = np.asarray(y)
    pred = (proba >= threshold).astype(int)
    return np.array([macro_f1(y[va], pred[va]) for _, va in folds])
```

```python
def impute(frames, columns, strategy='median'):
    fit = frames[0][columns]
    values = fit.median() if strategy == 'median' else fit.mode().iloc[0]
    for frame in frames:
        frame[columns] = frame[columns].fillna(values)
```

### Outliers

This part covers values far from the rest of the data.

`cap_outliers` memotong nilai di luar persentil `q` dan `1 - q` data fit. Kolom yang batas bawah dan atasnya sama tidak dipotong.

```python
def cap_outliers(frames, columns, q=0.005):
    lower, upper = frames[0][columns].quantile(q), frames[0][columns].quantile(1 - q)
    same = lower >= upper
    lower, upper = lower.where(~same, -np.inf), upper.where(~same, np.inf)
    for frame in frames:
        frame[columns] = frame[columns].clip(lower, upper, axis=1)
```

### Feature Scaling

This part covers normalization to the range 0 to 1 and standardization.

`scale` melakukan robust scaling (x - median) / IQR dengan statistik dari data fit. Kalau IQR 0, dipakai simpangan baku. `weight` lalu mengalikan kolom dengan bobot fitur.

```python
def scale(frames, columns):
    fit = frames[0][columns].astype(float)
    center = fit.median()
    spread = fit.quantile(0.75) - fit.quantile(0.25)
    spread = spread.where(spread > 0, fit.std()).replace(0, 1)
    for frame in frames:
        frame[columns] = (frame[columns].astype(float) - center) / spread

def weight(frames, weights):
    for frame in frames:
        for column, value in weights.items():
            frame[column] = frame[column] * value
```

### Encoding

This part covers the numeric representation of categorical attributes.

`encode` membuat kolom onehot dari level kategori di data fit. Nilai onehot `value` bisa lebih kecil dari 1.

```python
def encode(frames, columns, value=1.0):
    names = []
    for column in columns:
        for level in sorted(frames[0][column].dropna().unique()):
            for frame in frames:
                frame[f'{column}_{level}'] = (frame[column] == level) * value
            names.append(f'{column}_{level}')
    return names
```

# Preparation Pipeline <a name="6"></a>

The pipeline combines the preparation steps so they run in one consistent order.

```python
# preprocessing bersama untuk SEMUA model, dengan urutan:
# data error jadi missing, clipping, fitur turunan, imputasi, robust scaling, bobot fitur, onehot
# statistik diambil dari fit_df saja
# hasilnya matriks float32 untuk fit_df dan tiap df lain
def prepare(fit_df, *dfs):
    frames = [df.copy() for df in (fit_df,) + dfs]
    mark_invalid(frames)
    cap_outliers(frames, clip_col, CFG.CLIP_QUANTILE)
    engineer(frames)
    impute(frames, num_col, 'median')
    impute(frames, cat_col, 'most_frequent')
    scale(frames, num_col)
    weight(frames, CFG.FEATURE_WEIGHTS)
    columns = num_col + encode(frames, cat_col, CFG.ONEHOT_VALUE)
    prepare.columns = columns
    return [frame[columns].to_numpy(np.float32) for frame in frames]
```

## Logistic Regression

This part covers the model with L1, L2, and Elastic Net regularization.

`LogisticRegression` from scratch (bagian Implementasi From Scratch) mendukung regularisasi **L1, L2, dan Elastic Net**, bobot kelas, dan training di CPU atau GPU. Probabilitas batal untuk satu pemesanan dengan fitur $x$:

$$P(y = 1 \mid x) = \sigma(z), \qquad z = b + w^\top x, \qquad \sigma(z) = \frac{1}{1 + e^{-z}}$$

Fungsi objektif disamakan dengan scikit-learn supaya nilai `C` bisa dibandingkan langsung, lalu dibagi $C \cdot S$:

$$J(w, b) = \frac{1}{S}\sum_i s_i \big[\log(1 + e^{z_i}) - y_i z_i\big] + \frac{1 - \rho}{2\,C\,S}\|w\|_2^2 + \frac{\rho}{C\,S}\|w\|_1$$

Dengan $s_i$ bobot sampel, $S = \sum_i s_i$, $\rho = 0$ untuk L2, $\rho = 1$ untuk L1, dan $\rho$ = `l1_ratio` untuk Elastic Net. Bias tidak diregularisasi. Training memakai solver **FISTA** (gradient descent terakselerasi dengan proximal step untuk L1), yang dijelaskan di bagian Referensi [1], [2].

Prediksi 1 (batal) jika $P(y = 1 \mid x) \ge \tau$, dengan threshold $\tau$ dipilih dari OOF yang memberi Macro F1 tertinggi.

Ekspansi fitur di bawah dijalankan **setelah** `prepare` milik tim dan tidak mengubahnya. Langkahnya, semua dipelajari dari data fit saja:

1. Lima kolom numerik (`BIN_COLS`) dipotong menjadi bin dengan batas yang paling memisahkan kelas (pemisahan biner berulang berdasarkan impuritas Gini, minimal 200 baris per bin) lalu di-onehot.
2. Seluruh kolom (asli dan bin) distandardisasi.
3. Tujuh kolom (`INTER_COLS`) dikalikan dengan seluruh kolom lain yang bukan turunan fitur yang sama.
4. Hasil perkalian distandardisasi ulang, dan kolom yang konstan dibuang.

```python
# ekspansi fitur khusus LR di atas matriks prepare milik tim, semua statistik dipelajari dari data fit saja
class ExpandedPrep:
    def __init__(self, bin_cols, inter_cols, n_bins, min_leaf):
        self.bin_cols, self.inter_cols, self.n_bins, self.min_leaf = bin_cols, inter_cols, n_bins, min_leaf

    # tumpuk kolom asli dan kolom onehot bin
    def _stack(self, X):
        columns, names = [X[:, i] for i in range(X.shape[1])], list(self.base_)
        for c in self.bin_cols:
            bins = np.searchsorted(self.edges_[c], X[:, self.index_[c]], side='right')
            for k in range(1, len(self.edges_[c]) + 1):
                columns.append((bins == k).astype(float))
                names.append(f'{c}_bin{k}')
        return np.column_stack(columns), names

    # hasil kali tiap pasangan yang dipilih
    def _products(self, scaled):
        return np.column_stack([scaled[:, self.pos_[a]] * scaled[:, self.pos_[b]] for a, b in self.pairs_])

    def fit(self, df):
        self.fit_df = df.reset_index(drop=True)
        X = prepare(self.fit_df)[0].astype(float)
        self.base_ = list(prepare.columns)
        self.index_ = {name: i for i, name in enumerate(self.base_)}
        y = self.fit_df[CFG.TARGET].to_numpy()
        self.edges_ = {c: target_edges(X[:, self.index_[c]], y, self.n_bins, self.min_leaf) for c in self.bin_cols}

        stacked, names = self._stack(X)
        self.mean1_, self.std1_ = stacked.mean(axis=0), stacked.std(axis=0, ddof=1)
        self.keep1_ = self.std1_ > 0
        scaled = (stacked - self.mean1_) / np.where(self.keep1_, self.std1_, 1.0)

        # kolom dari fitur yang sama tidak dipasangkan dan tiap pasangan hanya dibuat sekali
        self.pos_ = {name: i for i, name in enumerate(names)}
        self.pairs_, seen = [], set()
        for a in self.inter_cols:
            for b, i in self.pos_.items():
                if not self.keep1_[i] or b == a or b.split('_bin')[0] == a or frozenset((a, b)) in seen:
                    continue
                seen.add(frozenset((a, b)))
                self.pairs_.append((a, b))
        products = self._products(scaled)
        self.mean2_, self.std2_ = products.mean(axis=0), products.std(axis=0, ddof=1)
        self.keep2_ = self.std2_ > 0
        self.names_ = ([n for n, k in zip(names, self.keep1_) if k]
                       + [f'{a}*{b}' for (a, b), k in zip(self.pairs_, self.keep2_) if k])
        return self

    def transform(self, df):
        X = prepare(self.fit_df, df)[1].astype(float)
        stacked, _ = self._stack(X)
        scaled = (stacked - self.mean1_) / np.where(self.keep1_, self.std1_, 1.0)
        products = (self._products(scaled) - self.mean2_) / np.where(self.keep2_, self.std2_, 1.0)
        return pd.DataFrame(np.hstack([scaled[:, self.keep1_], products[:, self.keep2_]]), columns=self.names_)

    def fit_transform(self, df):
        return self.fit(df).transform(df)


def make_prep():
    return ExpandedPrep(CFG.BIN_COLS, CFG.INTER_COLS, CFG.N_BINS, CFG.BIN_MIN_LEAF)
```

```python
# device dan dtype bisa ditimpa, parameter lain bisa diganti lewat overrides
def make_model(device=None, dtype=None, **overrides):
    params = {**CFG.LR_PARAMS, **overrides}
    return LogisticRegression(device=device or CFG.DEVICE, dtype=dtype or CFG.DTYPE, **params)


# kelompok sebuah kolom hasil preprocessing berdasarkan namanya
def column_group(name):
    if '*' in name:
        return 'interaksi'
    if '_bin' in name:
        return 'bin'
    return 'dasar dan one-hot'


# fitur untuk Logistic Regression, statistik dipelajari dari train_set saja lalu diterapkan ke val_set dan test
prep_lr = make_prep().fit(train_set)
F_train, F_val, F_test = prep_lr.transform(train_set), prep_lr.transform(val_set), prep_lr.transform(test_df)

display(pd.Series({'F_train': F_train.shape, 'F_val': F_val.shape, 'F_test': F_test.shape}).to_frame('bentuk'))
print('NaN train/val/test:', int(F_train.isna().sum().sum()), '/', int(F_val.isna().sum().sum()), '/', int(F_test.isna().sum().sum()))
display(pd.Series([column_group(c) for c in F_train.columns]).value_counts().to_frame('jumlah_kolom'))
```

#### Insights

> Ekspansi menambah ratusan kolom di atas 28 kolom preprocessing bersama.

- Kolom bin memberi satu bobot per kelompok nilai, dan kolom interaksi memberi pengaruh yang berubah menurut fitur lain.
- Preprocessing bersama tidak diubah. `ExpandedPrep` hanya memakai keluaran `prepare` dan statistiknya dipelajari dari data latih.

```python
folds = list(StratifiedKFold(n_splits=CFG.N_FOLDS, shuffle=True, random_state=CFG.SEED).split(train_clean, y_clean))


# ringkas hasil OOF dan jika threshold tidak diberikan, dipilih threshold dengan Macro F1 OOF tertinggi
def report(name, oof, threshold=None):
    if threshold is None:
        threshold, score = best_threshold(y_clean, oof)
    else:
        score = macro_f1(y_clean, (oof >= threshold).astype(int))
    per_fold = fold_scores(y_clean, oof, folds, threshold)
    return dict(model=name, threshold=threshold, oof_macro_f1=round(score, 4),
                fold_mean=round(float(per_fold.mean()), 4), fold_std=round(float(per_fold.std()), 4))


start = time.perf_counter()
oof = cross_val_proba(make_model, make_prep, train_clean, y_clean, folds)
print(f'waktu validasi silang {time.perf_counter() - start:.0f} detik (device {CFG.DEVICE}, dtype {CFG.DTYPE})')

res_best = report('LR', oof)
res_cfg = report('LR', oof, threshold=CFG.THRESHOLD)
thr_oof = res_best['threshold']
fold_f1 = fold_scores(y_clean, oof, folds, CFG.THRESHOLD)
print('\n======== OOF Logistic Regression ========')
print(f'Macro F1 OOF pada threshold {CFG.THRESHOLD}: {res_cfg["oof_macro_f1"]:.4f}   (diharapkan {CFG.EXPECTED_OOF}, selisih {res_cfg["oof_macro_f1"] - CFG.EXPECTED_OOF:+.4f})')
print(f'threshold terbaik OOF: {thr_oof}   (dipakai: {CFG.THRESHOLD})')
print(f'std per fold: {fold_f1.std():.4f}   (per fold: {np.round(fold_f1, 4).tolist()})')
pred_oof = (oof >= CFG.THRESHOLD).astype(int)
print('confusion matrix OOF [[TN, FP], [FN, TP]]')
print(confusion_matrix(y_clean, pred_oof))
print('F1 kelas 0, kelas 1:', f1_per_class(y_clean, pred_oof).round(4))
```

### Tuning Regularisasi dan `C`

Grid jenis regularisasi dan kekuatannya, dengan preprocessing dijalankan satu kali per fold lalu semua konfigurasi dilatih di matriks yang sama. Setiap konfigurasi dinilai dengan threshold terbaiknya sendiri dari OOF. Konfigurasi yang dipakai di Settings dipilih dari tabel ini.

```python
# validasi silang beberapa model sekaligus dengan satu preprocessor per fold
def cv_multi(makers):
    oofs = {name: np.zeros(len(y_clean)) for name in makers}
    stats_fit = {name: [] for name in makers}
    for k, (tr, va) in enumerate(folds, 1):
        prep_k = make_prep()
        X_tr = prep_k.fit_transform(train_clean.iloc[tr])
        X_va = prep_k.transform(train_clean.iloc[va])
        for name, make in makers.items():
            start = time.perf_counter()
            m = make().fit(X_tr, y_clean[tr])
            stats_fit[name].append(dict(detik=time.perf_counter() - start, iterasi=m.n_iter_, tidak_nol=int((m.coef_ != 0).sum())))
            oofs[name][va] = m.predict_proba(X_va)
        print(f'fold {k} selesai')
    return oofs, stats_fit


MAIN_LABEL = f"{CFG.LR_PARAMS['penalty']} C={CFG.LR_PARAMS['C']}"
grid_oof, grid_stats = {MAIN_LABEL: oof}, {}
if CFG.RUN_C_SWEEP:
    makers = {}
    for penalty in ['l1', 'l2', 'elasticnet']:
        for c_value in CFG.C_GRID:
            label = f'{penalty} C={c_value}'
            if label != MAIN_LABEL:
                makers[label] = (lambda penalty=penalty, c_value=c_value: make_model(penalty=penalty, C=c_value))
    start = time.perf_counter()
    new_oof, grid_stats = cv_multi(makers)
    grid_oof.update(new_oof)
    print(f'waktu grid {time.perf_counter() - start:.0f} detik untuk {len(makers)} konfigurasi')

rows = []
for label, p in grid_oof.items():
    row = report(label, p)
    st = grid_stats.get(label)
    row.update(tidak_nol=round(float(np.mean([s['tidak_nol'] for s in st])), 0) if st else '-',
               iterasi=round(float(np.mean([s['iterasi'] for s in st])), 0) if st else '-')
    rows.append(row)
grid_table = pd.DataFrame(rows).sort_values('oof_macro_f1', ascending=False).reset_index(drop=True)
display(grid_table)
```

```python
best = grid_table.iloc[0]
final_row = grid_table[grid_table['model'] == MAIN_LABEL].iloc[0]
by_penalty = grid_table.assign(penalty=grid_table['model'].str.split(' ').str[0]).groupby('penalty')['oof_macro_f1'].max()
spread = float(grid_table['oof_macro_f1'].max() - grid_table['oof_macro_f1'].min())
display(Markdown(
    f'#### Insights\n\n'
    f'> Konfigurasi dengan skor tertinggi pada grid adalah **{best["model"]}** ({best["oof_macro_f1"]:.4f}), dan yang dipakai di Settings adalah **{MAIN_LABEL}** ({final_row["oof_macro_f1"]:.4f}).\n\n'
    f'- Skor terbaik per jenis regularisasi: ' + ', '.join(f'{k} {v:.4f}' for k, v in by_penalty.items()) + '.\n'
    f'- Selisih skor terbesar antar konfigurasi pada grid adalah {spread:.4f}.\n'
    f'- Kolom `tidak_nol` menunjukkan jumlah kolom yang masih dipakai: bobot L1 dan Elastic Net bisa tepat nol, sedangkan L2 hanya mengecilkan bobot.\n'
    f'- Nilai `C` yang terlalu kecil membuat model underfit karena bobot terlalu dikecilkan, dan yang terlalu besar membuat regularisasi hampir tidak bekerja.'))
```

# Comparison with scikit-learn <a name="8"></a>

---

This section compares each from-scratch model with its `scikit-learn` counterpart.

Model from scratch dibandingkan dengan `sklearn.linear_model.LogisticRegression` pada **holdout** yang sama (`train_set` untuk training dan `val_set` untuk evaluasi, dengan matriks fitur yang sama persis). Fungsi objektif kedua model sama, sehingga `C` bisa dipakai langsung. `C = 0,1` dipakai untuk kedua pasangan.

| Pasangan | from scratch | scikit-learn |
| --- | --- | --- |
| L2 | FISTA | `lbfgs` |
| L1 | FISTA | `saga` (jalan jika `RUN_SK_L1 = True`) |

Metrik: nilai fungsi objektif di data latih (makin kecil makin dekat ke titik optimum), log-loss dan Macro F1 di `val_set` pada threshold model, selisih probabilitas antar model, dan waktu. Jika dijalankan di GPU, model L1 from scratch juga dilatih di CPU untuk memastikan hasilnya sama.

```python
# mulai scikit-learn 1.8 jenis regularisasi diatur lewat l1_ratio, versi lama memakai parameter penalty
SK_NEW_API = tuple(int(v) for v in sklearn.__version__.split('.')[:2]) >= (1, 8)


# buat LogisticRegression scikit-learn dengan fungsi objektif yang sama dengan model from scratch
# lbfgs dipakai untuk L2 dan saga untuk L1 karena lbfgs tidak mendukung L1
def make_sklearn(penalty, C, tol=None, max_iter=10000):
    ratio = {'l2': 0.0, 'l1': 1.0}[penalty]
    solver = 'lbfgs' if penalty == 'l2' else 'saga'
    tol = tol if tol is not None else (1e-10 if solver == 'lbfgs' else 1e-6)
    common = dict(C=C, solver=solver, tol=tol, max_iter=max_iter)
    if SK_NEW_API:
        return SkLogisticRegression(l1_ratio=ratio, **common)
    return SkLogisticRegression(penalty=penalty, **common)


# nilai fungsi objektif yang sama dengan model from scratch, dihitung dari bobot model mana pun
def objective(coef, intercept, X, yv, C, penalty):
    z = np.asarray(X, dtype=float) @ coef + intercept
    data_term = float(np.mean(np.logaddexp(0.0, z) - yv * z))
    penalty_term = float(np.abs(coef).sum()) if penalty == 'l1' else 0.5 * float(coef @ coef)
    return data_term + penalty_term / (C * len(yv))


rows, probs = [], {}


# catat satu baris hasil perbandingan
def add_row(label, penalty, C, coef, intercept, proba, seconds, iterations):
    rows.append(dict(model=label, waktu_detik=round(seconds, 1), iterasi=iterations,
                     bobot_tidak_nol=int((coef != 0).sum()),
                     objektif_train=round(objective(coef, intercept, F_train, y_train, C, penalty), 6),
                     logloss_val=round(float(log_loss(y_val, proba)), 5),
                     macro_f1_val=round(macro_f1(y_val, (proba >= CFG.THRESHOLD).astype(int)), 4)))
    probs[label] = proba


# latih model scikit-learn dan peringatan konvergensi diredam karena jumlah iterasi dilaporkan di tabel
def fit_sklearn(label, penalty, C, **kwargs):
    sk = make_sklearn(penalty, C, **kwargs)
    start = time.perf_counter()
    with warnings.catch_warnings():
        warnings.simplefilter('ignore', ConvergenceWarning)
        sk.fit(F_train, y_train)
    seconds = time.perf_counter() - start
    add_row(label, penalty, C, sk.coef_[0], float(sk.intercept_[0]), sk.predict_proba(F_val)[:, 1], seconds, int(np.max(sk.n_iter_)))
    return sk


# model from scratch di holdout sudah dilatih di bagian Modeling dengan C yang sama
C_SK = CFG.LR_PARAMS['C']
dev = CFG.DEVICE.upper()
times = {r['regularisasi']: r['waktu_detik'] for r in holdout_rows}
for penalty, label in [('l2', 'L2'), ('l1', 'L1')]:
    m = holdout_models[penalty]
    add_row(f'scratch {label} ({dev})', penalty, C_SK, m.coef_, m.intercept_, m.predict_proba(F_val), times[label], m.n_iter_)

print('L2 scikit-learn (lbfgs) ...')
sk_l2 = fit_sklearn('sklearn L2 (lbfgs)', 'l2', C_SK, tol=1e-8, max_iter=3000)
if CFG.RUN_SK_L1:
    print('L1 scikit-learn (saga), bisa memakan beberapa menit ...')
    sk_l1 = fit_sklearn('sklearn L1 (saga)', 'l1', C_SK, tol=1e-4, max_iter=1000)
if CFG.RUN_CPU_CHECK and CFG.DEVICE != 'cpu':
    print('L1 from scratch di CPU ...')
    start = time.perf_counter()
    m_cpu = make_model(device='cpu', penalty='l1').fit(F_train, y_train)
    add_row('scratch L1 (CPU)', 'l1', C_SK, m_cpu.coef_, m_cpu.intercept_, m_cpu.predict_proba(F_val), time.perf_counter() - start, m_cpu.n_iter_)

# selisih probabilitas validasi terhadap model from scratch dengan regularisasi yang sama
ref = {'L1': f'scratch L1 ({dev})', 'L2': f'scratch L2 ({dev})'}
table = pd.DataFrame(rows)
table['selisih_proba_maks'] = [float(np.abs(probs[m] - probs[ref['L1' if 'L1' in m else 'L2']]).max()) for m in table['model']]
table['selisih_proba_rata2'] = [float(np.abs(probs[m] - probs[ref['L1' if 'L1' in m else 'L2']]).mean()) for m in table['model']]
display(table.set_index('model').round(6))
```

```python
def objective_gap(a, b):
    ra = table.set_index('model')['objektif_train']
    return abs(float(ra[a] - ra[b])) if a in ra.index and b in ra.index else None


gap_l2 = objective_gap(f'scratch L2 ({dev})', 'sklearn L2 (lbfgs)')
gap_l1 = objective_gap(f'scratch L1 ({dev})', 'sklearn L1 (saga)')
lines = []
if gap_l2 is not None:
    lines.append(f'Pada L2, selisih nilai objektif antara from scratch dan scikit-learn (`lbfgs`) hanya {gap_l2:.2e}, '
                 f'dengan selisih probabilitas validasi paling besar {table.loc[table["model"] == "sklearn L2 (lbfgs)", "selisih_proba_maks"].iloc[0]:.4f}.')
if gap_l1 is not None:
    lines.append(f'Pada L1, selisih nilai objektif terhadap scikit-learn (`saga`) adalah {gap_l1:.2e}. '
                 f'Toleransi `saga` dibuat longgar (1e-4, maksimal 1000 epoch) supaya selesai dalam waktu wajar, jadi nilai objektif yang lebih kecil berarti lebih dekat ke titik optimum.')
else:
    lines.append('Pasangan L1 dengan `saga` dilewati (`RUN_SK_L1 = False`, bawaan di CPU). Ubah di Settings untuk menjalankannya.')
if CFG.RUN_CPU_CHECK and CFG.DEVICE != 'cpu':
    diff_pred = int(((probs[f'scratch L1 ({dev})'] >= CFG.THRESHOLD) != (probs['scratch L1 (CPU)'] >= CFG.THRESHOLD)).sum())
    lines.append(f'GPU dan CPU menghasilkan model yang sama sampai ketelitian numerik, dengan {diff_pred} prediksi berbeda dari {len(y_val)} baris validasi.')
display(Markdown('#### Insights\n\n> Model from scratch menghasilkan model yang setara dengan scikit-learn.\n\n' + '\n'.join('- ' + s for s in lines)))
```

## Cost-Sensitive Classification

Pada task ini kelas positif adalah pemesanan yang batal.

- **False Positive (FP)**: pemesanan yang sebenarnya datang diprediksi batal. Hotel menjual ulang kamarnya, sehingga saat tamu datang kamar sudah penuh (overbooked).
- **False Negative (FN)**: pemesanan yang batal tidak terdeteksi, sehingga kamar kosong dan pendapatan hilang.

Dengan asumsi FP dua kali lebih mahal daripada FN:

| | Prediksi tidak batal | Prediksi batal |
| --- | --- | --- |
| **Aktual tidak batal** | benar, biaya 0 | **FP, biaya 2** |
| **Aktual batal** | **FN, biaya 1** | benar, biaya 0 |

Metrik yang diminimalkan adalah **biaya total = 2 x FP + FN**.

**Threshold optimal.** Jika model memberi probabilitas batal $p$ yang terkalibrasi, biaya harapan prediksi batal adalah $(1-p)\,c_{FP}$ dan biaya harapan prediksi tidak batal adalah $p\,c_{FN}$. Prediksi batal lebih murah jika

$$p > \frac{c_{FP}}{c_{FP} + c_{FN}} = \frac{2}{2 + 1} = \frac{2}{3}$$

Jadi threshold naik dari sekitar 0,5 menjadi sekitar 0,67. Model menjadi lebih hati-hati menyatakan sebuah pemesanan batal.

**Pembobotan kelas.** Cara kedua adalah memasukkan biaya ke training. Sampel kelas 0 (sumber FP) diberi bobot 2 pada log-loss, sehingga threshold teoretisnya kembali ke 0,5. Kedua cara ini dibahas oleh Elkan [3]. Keduanya dibandingkan dengan threshold yang dipilih langsung dari data (biaya OOF terkecil).

```python
COST_FP, COST_FN = CFG.COST_FP, CFG.COST_FN
THR_GRID = np.round(np.arange(0.05, 0.951, 0.005), 3)


# biaya total dengan FP lebih mahal dari FN
def total_cost(y_true, y_pred):
    (tn, fp), (fn, tp) = confusion_matrix(y_true, y_pred)
    return COST_FP * fp + COST_FN * fn


# threshold dengan biaya OOF terkecil
def min_cost_threshold(proba):
    costs = np.array([total_cost(y_clean, (proba >= t).astype(int)) for t in THR_GRID])
    return float(THR_GRID[int(np.argmin(costs))])


# ringkasan satu strategi: FP, FN, biaya, Macro F1, serta presisi dan recall kelas batal
def evaluate(label, proba, threshold):
    pred = (proba >= threshold).astype(int)
    (tn, fp), (fn, tp) = confusion_matrix(y_clean, pred)
    return dict(strategi=label, threshold=round(threshold, 3), FP=fp, FN=fn,
                biaya_total=COST_FP * fp + COST_FN * fn,
                biaya_per_pemesanan=round((COST_FP * fp + COST_FN * fn) / len(y_clean), 4),
                macro_f1=round(macro_f1(y_clean, pred), 4),
                presisi_batal=round(tp / max(tp + fp, 1), 4), recall_batal=round(tp / max(tp + fn, 1), 4))


# model dengan bobot kelas 2:1 dilatih pada fitur yang sama, jumlah fold dan preprocessing sama dengan model biasa
make_model_weighted = lambda: make_model(class_weight={0: COST_FP, 1: COST_FN})
start = time.perf_counter()
oof_w = cross_val_proba(make_model_weighted, make_prep, train_clean, y_clean, folds)
print(f'waktu validasi model berbobot {time.perf_counter() - start:.0f} detik')
print('rata-rata probabilitas tanpa bobot', round(oof.mean(), 4), '| dengan bobot', round(oof_w.mean(), 4))

thr_f1 = CFG.THRESHOLD
thr_cost = min_cost_threshold(oof)
thr_cost_w = min_cost_threshold(oof_w)
```

```python
cost_plain = [total_cost(y_clean, (oof >= t).astype(int)) / len(y_clean) for t in THR_GRID]
cost_w = [total_cost(y_clean, (oof_w >= t).astype(int)) / len(y_clean) for t in THR_GRID]
f1_plain = [macro_f1(y_clean, (oof >= t).astype(int)) for t in THR_GRID]

figure, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].plot(THR_GRID, cost_plain, label='tanpa bobot')
axes[0].plot(THR_GRID, cost_w, label='bobot kelas 2:1')
axes[0].axvline(2 / 3, color='gray', linestyle=':', label='teori 2/3')
axes[0].axvline(thr_cost, color='tab:blue', linestyle='--', label=f'minimum tanpa bobot {thr_cost}')
axes[0].axvline(thr_cost_w, color='tab:orange', linestyle='--', label=f'minimum dengan bobot {thr_cost_w}')
axes[0].set_xlabel('threshold')
axes[0].set_ylabel('biaya per pemesanan')
axes[0].set_title('Biaya OOF (2 FP + FN)')
axes[0].legend(fontsize=8)
axes[1].plot(THR_GRID, f1_plain, color='tab:green')
axes[1].axvline(thr_f1, color='tab:green', linestyle='--', label=f'optimal Macro F1 {thr_f1}')
axes[1].axvline(thr_cost, color='tab:blue', linestyle='--', label=f'optimal biaya {thr_cost}')
axes[1].set_xlabel('threshold')
axes[1].set_ylabel('Macro F1 OOF')
axes[1].set_title('Macro F1 pada model tanpa bobot')
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()
```

```python
strategies = [
    ('tanpa model: semua tidak batal',          np.zeros(len(y_clean)), 0.5),
    ('tanpa penanganan: threshold Macro F1',    oof, thr_f1),
    ('threshold default 0,5',                   oof, 0.5),
    ('threshold teoretis 2/3',                  oof, 2 / 3),
    ('threshold empiris biaya minimum',         oof, thr_cost),
    ('bobot kelas 2:1, threshold 0,5',          oof_w, 0.5),
    ('bobot kelas 2:1, threshold empiris',      oof_w, thr_cost_w),
]
cost_table = pd.DataFrame([evaluate(label, p, t) for label, p, t in strategies])
base_cost = cost_table.loc[1, 'biaya_total']
cost_table['hemat_vs_tanpa_penanganan'] = ((base_cost - cost_table['biaya_total']) / base_cost * 100).round(1).astype(str) + '%'
display(cost_table)

models_only = cost_table.iloc[1:]
best_cost = models_only.loc[models_only['biaya_total'].idxmin()]
base = cost_table.iloc[1]
display(Markdown(
    f'#### Insights\n\n'
    f'> Strategi dengan biaya terkecil adalah **{best_cost["strategi"]}** (threshold {best_cost["threshold"]}), biaya total {int(best_cost["biaya_total"]):,} '
    f'({best_cost["hemat_vs_tanpa_penanganan"]} lebih hemat).\n\n'
    f'- Tanpa penanganan (threshold Macro F1 {thr_f1}), biaya total {int(base["biaya_total"]):,} dengan {int(base["FP"]):,} FP dan {int(base["FN"]):,} FN.\n'
    f'- Pada strategi terbaik, FP turun menjadi {int(best_cost["FP"]):,} dan FN naik menjadi {int(best_cost["FN"]):,}. Model sengaja membiarkan lebih banyak batal yang tidak terdeteksi demi menghindari overbooked.\n'
    f'- Harganya adalah Macro F1, dari {base["macro_f1"]:.4f} menjadi {best_cost["macro_f1"]:.4f}. Ini trade-off yang diharapkan karena Macro F1 memperlakukan FP dan FN sama.\n'
    f'- Threshold teoretis 2/3 memberi biaya {int(cost_table.loc[3, "biaya_total"]):,} dan threshold empiris dari data memberi {int(cost_table.loc[4, "biaya_total"]):,}. '
    f'Keduanya berdekatan jika probabilitas model cukup terkalibrasi.\n'
    f'- Pembobotan kelas 2:1 (threshold 0,5) memberi biaya {int(cost_table.loc[5, "biaya_total"]):,}, mirip dengan menaikkan threshold, karena kedua cara memindahkan batas keputusan ke arah yang sama.'))
```

**Penanganan yang berubah.** Keputusan tidak lagi dinilai dengan Macro F1 saja, tetapi dengan biaya total $2\,FP + FN$. Threshold naik dari threshold Macro F1 ke kisaran 0,6 sampai 0,7 menurut teori, dan nilai persisnya dari data ada pada tabel di atas. Alternatifnya memberi bobot 2 pada kelas 0 saat training. Skor kompetisi tetap Macro F1, sehingga submission Kaggle memakai threshold Macro F1. Strategi cost-sensitive adalah rekomendasi untuk operasional hotel, bukan untuk leaderboard.

## Optimasi Model dengan FISTA

Gradient descent biasa (diajarkan di kelas) menghitung gradien di titik parameter sekarang. FISTA menghitung gradien di titik yang sudah digeser searah gerakan sebelumnya (momentum Nesterov), ditambah adaptive restart dan langkah $1/L$. Kedua solver meminimalkan fungsi objektif yang sama, jadi model akhirnya sama dan FISTA hanya membuat training lebih cepat. Perbandingan di bawah memakai fitur dan konfigurasi notebook ini pada `train_set`. Bagian gradient descent biasa lambat, jadi hanya berjalan jika `RUN_OPTIMIZER_GD = True`.

```python
# latih satu konfigurasi pada train_set dan catat iterasi, waktu, serta nilai objektif akhir
def benchmark(label, **params):
    start = time.perf_counter()
    m = make_model(**params).fit(F_train, y_train)
    return m, dict(solver=label, iterasi=m.n_iter_, konvergen=bool(m.converged_),
                   waktu_detik=round(time.perf_counter() - start, 2), objektif=round(m.loss_history_[-1], 7),
                   bobot_tidak_nol=int((m.coef_ != 0).sum()))


# gambar selisih objektif terhadap nilai terbaik di setiap iterasi dalam skala log
def plot_gap(models, title):
    best_obj = min(m.loss_history_[-1] for m in models.values())
    plt.figure(figsize=(7, 3.5))
    for label, m in models.items():
        gap = np.maximum(np.array(m.loss_history_) - best_obj, 1e-12)
        plt.semilogy(gap, label=label)
    plt.xlabel('iterasi')
    plt.ylabel('J(theta) - J terbaik')
    plt.title(title)
    plt.legend()
    plt.show()


opt_models, opt_rows = {}, []
m_fista, r_fista = benchmark('FISTA', solver='fista')
opt_models['FISTA'] = m_fista
opt_rows.append(r_fista)
if CFG.RUN_OPTIMIZER_GD:
    m_gd, r_gd = benchmark('GD', solver='gd', max_iter=3000)
    opt_models['GD'] = m_gd
    opt_rows.append(r_gd)
display(pd.DataFrame(opt_rows))
plot_gap(opt_models, f'Fitur final ({F_train.shape[1]} kolom), {CFG.LR_PARAMS["penalty"].upper()} C={CFG.LR_PARAMS["C"]}')
```

```python
if 'GD' in opt_models:
    ratio_iter = r_gd['iterasi'] / r_fista['iterasi']
    ratio_time = r_gd['waktu_detik'] / max(r_fista['waktu_detik'], 1e-9)
    gd_state = 'konvergen' if r_gd['konvergen'] else f'belum memenuhi toleransi setelah {r_gd["iterasi"]:,} iterasi'
    gap = r_gd['objektif'] - r_fista['objektif']
    p_gap = float(np.abs(opt_models['GD'].predict_proba(F_val) - m_fista.predict_proba(F_val)).max())
    display(Markdown(
        f'#### Insights\n\n'
        f'> FISTA konvergen dalam {r_fista["iterasi"]:,} iterasi ({r_fista["waktu_detik"]} detik), sedangkan gradient descent biasa {gd_state} ({r_gd["waktu_detik"]} detik).\n\n'
        f'- Gradient descent memakai {ratio_iter:.1f} kali lebih banyak iterasi dan {ratio_time:.1f} kali lebih banyak waktu.\n'
        f'- Nilai objektif GD masih {gap:+.2e} terhadap FISTA, dan probabilitas validasi keduanya berbeda paling besar {p_gap:.4f}.\n'
        f'- Pada grafik, galat FISTA turun beberapa orde lebih cepat, dan lekukan pada kurvanya adalah titik ketika adaptive restart me-reset momentum.'))
else:
    display(Markdown('#### Insights\n\n> Bagian gradient descent dilewati (`RUN_OPTIMIZER_GD = False`, bawaan di CPU), sehingga hanya FISTA yang ditampilkan.'))
```

## Video Kontur Log-Loss dan Lintasan Parameter

Agar fungsi loss bisa digambar dalam dua dimensi, model dibuat hanya dengan dua parameter:

$$P(\text{batal}) = \sigma(\theta_0 + \theta_1 x)$$

dengan $x$ adalah `special_requests` yang sudah distandardisasi, yaitu sinyal tunggal terkuat pada data. Regularisasi tidak dipakai agar yang digambar murni log-loss:

$$J(\theta_0, \theta_1) = \frac{1}{m} \sum_i \Big[\log\big(1 + e^{z_i}\big) - y_i z_i\Big], \quad z_i = \theta_0 + \theta_1 x_i$$

Tiga lintasan dibandingkan dari titik awal yang sama (0, 0):

| Lintasan | Keterangan |
| --- | --- |
| GD, learning rate 0,5 | gradient descent seperti di kelas, langkah kecil dan stabil |
| GD, learning rate 7 | langkah terlalu besar, parameter melompat ke sisi lain lembah |
| FISTA | gradient descent dengan momentum Nesterov dan langkah 1/L |

Training dua parameter ini berjalan di CPU karena terlalu kecil untuk GPU. Gambar dan video disimpan di `output/figures/`.

```python
# special_requests distandardisasi (z-score) dengan median untuk missing, sama dengan notebook LR lain
x_raw = train_clean[CFG.CONTOUR_FEATURE].fillna(train_clean[CFG.CONTOUR_FEATURE].median())
x_contour = pd.DataFrame({CFG.CONTOUR_FEATURE: (x_raw - x_raw.mean()) / x_raw.std()})

common = dict(penalty=None, tol=1e-6, max_iter=5000, record_params=True)
runs = {
    'GD lr=0,5': LogisticRegression(solver='gd', learning_rate=0.5, **common).fit(x_contour, y_clean),
    'GD lr=7': LogisticRegression(solver='gd', learning_rate=7.0, **common).fit(x_contour, y_clean),
    'FISTA (lr=1/L)': LogisticRegression(solver='fista', **common).fit(x_contour, y_clean),
}
for label, m in runs.items():
    print(f'{label:16s} iterasi {m.n_iter_:4d}  learning rate {m.learning_rate_:.2f}  '
          f'theta0 {m.intercept_:+.4f}  theta1 {m.coef_[0]:+.4f}  loss {m.loss_history_[-1]:.5f}')

final_params = np.array([[m.intercept_, m.coef_[0]] for m in runs.values()])
loss_final = np.array([m.loss_history_[-1] for m in runs.values()])
display(Markdown(
    f'#### Insights\n\n'
    f'> Ketiga lintasan berakhir di titik yang sama (selisih parameter terbesar {np.abs(final_params - final_params[0]).max():.1e}, selisih log-loss {loss_final.max() - loss_final.min():.1e}).\n\n'
    f'- Ini sesuai sifat log-loss yang cembung: hanya ada satu titik minimum, dan learning rate hanya memengaruhi jalan menuju titik itu.\n'
    f'- Jumlah iterasi masing-masing: ' + ', '.join(f'{k} {m.n_iter_}' for k, m in runs.items()) + '.'))
```

```python
xv = x_contour[CFG.CONTOUR_FEATURE].to_numpy()
values, inverse = np.unique(xv, return_inverse=True)
n_pos = np.bincount(inverse, weights=y_clean, minlength=len(values))
n_neg = np.bincount(inverse, minlength=len(values)) - n_pos


# log-loss rata-rata untuk setiap pasangan (theta0, theta1) pada grid, data diringkas per nilai unik
def loss_surface(T0, T1):
    Z = T0[..., None] + T1[..., None] * values
    softplus = np.logaddexp(0.0, Z)
    return (n_pos * (softplus - Z) + n_neg * softplus).sum(-1) / len(y_clean)


paths = np.vstack([m.param_history_ for m in runs.values()])
lo, hi = paths.min(0), paths.max(0)
pad = 0.15 * (hi - lo) + 0.2
T0, T1 = np.meshgrid(np.linspace(lo[0] - pad[0], hi[0] + pad[0], 300),
                     np.linspace(lo[1] - pad[1], hi[1] + pad[1], 300))
L = loss_surface(T0, T1)

m_ref = runs['FISTA (lr=1/L)']
check = loss_surface(np.array([[m_ref.intercept_]]), np.array([[m_ref.coef_[0]]]))[0, 0]
print('loss di titik optimum dari grid', round(check, 6), '| dari model', round(m_ref.loss_history_[-1], 6))
assert np.isclose(check, m_ref.loss_history_[-1])

levels = L.min() + np.geomspace(1e-4, L.max() - L.min(), 25)
colors = {'GD lr=0,5': 'tab:blue', 'GD lr=7': 'tab:red', 'FISTA (lr=1/L)': 'tab:green'}

figure, ax = plt.subplots(figsize=(7, 6))
cs = ax.contour(T0, T1, L, levels=levels, cmap='viridis', linewidths=0.8)
ax.clabel(cs, cs.levels[::4], fmt='%.3f', fontsize=7)
for label, m in runs.items():
    h = m.param_history_
    ax.plot(h[:, 0], h[:, 1], marker='o', markersize=2.5, linewidth=1, color=colors[label], label=f'{label}, {m.n_iter_} iterasi')
ax.plot(m_ref.intercept_, m_ref.coef_[0], marker='*', markersize=14, color='black', label='optimum')
ax.set_xlabel('theta0 (bias)')
ax.set_ylabel(f'theta1 (bobot {CFG.CONTOUR_FEATURE})')
ax.set_title('Kontur log-loss dan lintasan parameter')
ax.legend(fontsize=8)
static_path = CFG.FIG_DIR / f'{CFG.VIDEO_NAME}.png'
figure.savefig(static_path, dpi=130, bbox_inches='tight')
plt.show()
print('gambar tersimpan di', static_path.relative_to(ROOT))
```

## Metode pada Model

**FISTA (Fast Iterative Shrinkage-Thresholding Algorithm).** Gradient descent terakselerasi dengan momentum Nesterov dan proximal step. Gradien dihitung di titik yang sudah digeser searah gerakan sebelumnya:

$$v_k = \theta_k + \beta_k(\theta_k - \theta_{k-1}), \qquad \theta_{k+1} = \text{prox}\big(v_k - \eta \nabla J(v_k)\big)$$

$$t_{k+1} = \frac{1 + \sqrt{1 + 4t_k^2}}{2}, \qquad \beta_k = \frac{t_k - 1}{t_{k+1}}$$

- *Kenapa dipakai.* Matriks fitur berisi ribuan kolom interaksi yang saling berkorelasi, sehingga permukaan loss berbentuk lembah sempit dan panjang. Gradient descent biasa memantul di dinding lembah dan hanya maju sedikit. Momentum mengumpulkan arah yang konsisten. Untuk fungsi cembung, galat gradient descent turun sebanding $1/k$ dan FISTA sebanding $1/k^2$.
- *Kode.* `LogisticRegression.fit` dengan `solver='fista'` (bawaan). Perbandingan dengan gradient descent biasa ada di bagian Bonus.
- *Sumber.* [2], [1].

**Adaptive restart.** Jika arah langkah berbalik terhadap momentum, yaitu $(v_k - \theta_{k+1})^\top(\theta_{k+1} - \theta_k) > 0$, momentum di-reset ($t = 1$). Reset ini mencegah parameter berosilasi di sekitar titik optimum. Sumber: [4]. Kode: blok `if self.solver == 'fista'` di `LogisticRegression.fit`.

**Learning rate $1/L$.** Turunan kedua sigmoid paling besar 0,25, sehingga kelengkungan log-loss dibatasi $L \le 0{,}25\,\lambda_{\max}(X^\top S X / \sum s) + \lambda_2$. Langkah $1/L$ menjamin loss turun tanpa mencoba learning rate satu per satu. Kode: `LogisticRegression._step_size`. Sumber konsep konstanta Lipschitz: [1].

**Proximal step untuk L1 (soft-thresholding).** Suku L1 tidak terdiferensiasi di nol, sehingga setelah langkah gradien dipakai $\text{sign}(w)\max(|w| - \eta\lambda_1, 0)$. Bobot yang kecil menjadi tepat nol, dan inilah yang menjadi seleksi fitur otomatis pada L1. Kode: blok `if lam1 > 0` di `LogisticRegression.fit`. Sumber: [5], [6].

**Regularisasi L1, L2, dan Elastic Net.** L2 (ridge) mengecilkan bobot [7], L1 (lasso) membuat sebagian bobot nol [6], dan Elastic Net mencampur keduanya dengan porsi `l1_ratio` [8]. Fungsi objektif disamakan dengan scikit-learn supaya `C` bisa dibandingkan langsung.

**Bobot kelas untuk cost-sensitive.** Setiap sampel kelas tertentu dikalikan bobot pada log-loss, sehingga kesalahan yang lebih mahal ikut diperhitungkan saat training. Kode: parameter `class_weight` di `LogisticRegression`. Sumber: [3].

**Backend GPU (opsional).** Matriks disalin ke GPU sekali di awal dan seluruh iterasi (`X @ w` dan `X.T @ err`) berjalan di sana lewat CuPy, pustaka array dengan sintaks seperti NumPy. Hasil disalin kembali ke CPU, jadi model yang disimpan tetap bisa dipakai tanpa GPU. Sumber: [9].

## Metode pada Ekspansi Fitur

**Binning berbasis target.** Batas bin dicari dengan pemisahan biner berulang yang paling menurunkan impuritas Gini terhadap target (best-first, minimal 200 baris per bin), diadaptasi dari kriteria pemisahan CART [10]. Setiap bin di-onehot sehingga model memberi satu bobot per kelompok nilai. Batas bin dipelajari dari data latih saja. Kode: fungsi `target_edges` dan `ExpandedPrep`.

**Fitur interaksi.** Tujuh kolom terpilih dikalikan dengan seluruh kolom lain (kecuali yang berasal dari fitur yang sama), lalu hasilnya distandardisasi ulang. Model linear jadi bisa menangkap pengaruh yang saling bergantung, misalnya lead time yang berbeda efeknya pada tiap tahun. Kode: `ExpandedPrep.fit`.
